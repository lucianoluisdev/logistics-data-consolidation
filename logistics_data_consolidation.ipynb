```python
import pandas as pd
import os

# Pasta onde estão os arquivos
pasta = r'C:\Logistica'

# Lista para guardar dados
lista_planilhas = []

# Ler todos arquivos da pasta
for arquivo in os.listdir(pasta):

    if arquivo.endswith('.xlsx') and arquivo.startswith('Filial'):

        caminho = os.path.join(pasta, arquivo)

        df = pd.read_excel(caminho)

        # Adiciona nome do arquivo (origem)
        df['origem'] = arquivo

        lista_planilhas.append(df)

# Junta tudo
relatorio_final = pd.concat(lista_planilhas, ignore_index=True)

# Salvar arquivo final
saida = os.path.join(pasta, 'Relatorio_Final.xlsx')

relatorio_final.to_excel(saida, index=False)

print('Relatório consolidado criado com sucesso!')
print(saida)
```

import pandas as pd
import os
import random

# Pasta onde serão salvas
pasta = r'C:\Logistica'

os.makedirs(pasta, exist_ok=True)

# Lista de filiais reais
filiais = [
    'Sao_Paulo',
    'Campinas',
    'Santos',
    'Guarulhos',
    'Osasco',
    'Sorocaba',
    'Ribeirao_Preto',
    'Sao_Jose_dos_Campos',
    'Jundiai',
    'Bauru',
    'Piracicaba',
    'Santo_Andre',
    'Sao_Bernardo',
    'Diadema',
    'Mogi_das_Cruzes',
    'Barueri',
    'Cotia',
    'Taubate',
    'Americana',
    'Limeira'
]

# Dados fictícios
clientes = ['Loja A', 'Loja B', 'Loja C', 'Loja D']
transportadoras = ['Rápida', 'Expresso', 'Total', 'Flash']
status_lista = ['Entregue', 'Em Rota', 'Atrasado']

# Criar arquivos
for filial in filiais:

    dados = []

    for i in range(15):

        dados.append({
            'pedido': random.randint(1000, 9999),
            'cliente': random.choice(clientes),
            'cidade': filial.replace('_', ' '),
            'transportadora': random.choice(transportadoras),
            'status': random.choice(status_lista),
            'data_entrega': '28/04/2026'
        })

    df = pd.DataFrame(dados)

    arquivo = os.path.join(pasta, f'Filial_{filial}.xlsx')

    df.to_excel(arquivo, index=False)

print('20 planilhas criadas com sucesso em:')
print(pasta)
